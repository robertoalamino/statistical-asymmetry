# Figure 2 — Initial state for the cellular automata

The figure displays the 100-cell initial sequence in four consecutive
rows of 25 cells. Read each row from left to right, starting at the top.
Grey denotes +1 and white denotes −1. The code stores these states as
1 and 0, respectively.

NumPy's `RandomState(55)` reproduces all 100 values shown in the paper.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap

initial_state = np.random.RandomState(55).randint(0, 2, 100)
rows = initial_state.reshape(4, 25)

fig, ax = plt.subplots(figsize=(10, 2))
ax.pcolormesh(
    rows,
    cmap=ListedColormap(["white", "#606060"]),
    vmin=0,
    vmax=1,
    edgecolors="black",
    linewidth=1,
)
ax.invert_yaxis()
ax.set_aspect("equal")
ax.set_axis_off()
plt.show()
```
